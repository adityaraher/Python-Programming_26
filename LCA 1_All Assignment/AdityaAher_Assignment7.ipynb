Load a CSV dataset and plot it using line, bar, scatter and heatmap plots (Matplotlib and Seaborn).

```python
#Import the required libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#Load the dataset
df = pd.read_csv("sample_data_assignment_07.csv")
print(df)
```

```python
#Explore the dataset
print("Shape:", df.shape)
print()
df.info()
print()
print(df.describe())
```

```python
#Line Plot - Python marks of each student
plt.figure(figsize=(7, 4.5))
plt.plot(df["Student"], df["Python"], marker="o")
plt.title("Python Marks of Students")
plt.xlabel("Student")
plt.ylabel("Python Marks")
plt.grid(True)
plt.show()
```

```python
#Bar Plot - Maths marks of each student
plt.figure(figsize=(7, 4.5))
plt.bar(df["Student"], df["Maths"])
plt.title("Maths Marks of Students")
plt.xlabel("Student")
plt.ylabel("Maths Marks")
plt.show()
```

```python
#Scatter Plot - Maths vs Python marks
plt.figure(figsize=(7, 4.5))
plt.scatter(df["Maths"], df["Python"], s=70)
plt.title("Maths vs Python Marks")
plt.xlabel("Maths Marks")
plt.ylabel("Python Marks")
plt.grid(True)
plt.show()
```

```python
#Correlation matrix of Maths, Python and Attendance
correlation = df[["Maths", "Python", "Attendance"]].corr()
print("Correlation Matrix:")
print(correlation)
```

```python
#Heatmap of the correlation matrix
plt.figure(figsize=(6.5, 5))
sns.heatmap(correlation, annot=True, cmap="Blues", vmin=-1, vmax=1)
plt.title("Correlation Heatmap")
plt.show()
```
