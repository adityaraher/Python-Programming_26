```python
#dataframes are 2 dimensional data structures
import pandas as pd
data={
    "Name":["A","B","C"],
    "Age":[19,22,23],
    "Marks":[90,80,70],
    "Height":[175,180,155],
    "Weight":[67,88,55],
    "Division":[2,3,1]
}
df = pd.DataFrame(data)
print(df)
print(df["Age"])
print(df["Name"])
print(df[["Age","Name"]])#for simultaneously printing the column data types
print(df["Height"])
print(df["Weight"])
print(df.shape)
print(df.columns)
print(df.head())
print(df.tail())
print(df.head(2))
print(df.tail(2))
```
