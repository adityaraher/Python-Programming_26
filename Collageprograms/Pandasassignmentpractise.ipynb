```python
"""
Create a DataFrame with 8 rows and 4 columns of your choice. Write Python
statements to display:
1. The first 3 rows
2. The last 2 rows
3. The shape of the DataFrame
4. The column names
"""
import pandas as pd

data = {
    "ID": [101, 102, 103, 104, 105, 106, 107, 108],
    "City": ["Mumbai", "Delhi", "Bangalore", "Pune", "Chennai", "Kolkata",
             "Hyderabad", "Ahmedabad"],
    "Temperature": [32, 38, 28, 30, 35, 33, 34, 40],
    "Humidity": [80, 50, 65, 70, 75, 82, 60, 45]
}

df = pd.DataFrame(data)

# 1. The first 3 rows
print("1. First 3 rows:\n", df.head(3))

# 2. The last 2 rows
print("\n2. Last 2 rows:\n", df.tail(2))

# 3. The shape of the DataFrame
print("\n3. Shape:", df.shape)

# 4. The column names
print("\n4. Columns:", df.columns)
```

```python
"""
Consider the following DataFrame:
data = {
"Name": ["A", "B", "C", "D", "E", "F"],
"Age": [18, 19, 18, 20, 19, 18],
"Marks": [72, 85, 91, 68, 79, 95]
}
df = pd.DataFrame(data)
Write Python statements to display:
1. The first 2 rows using head()
2. The last 2 rows using tail()
3. The number of rows and columns using shape
4. The column names using columns
5. The data types using dtypes
6. The statistical summary using describe()
"""
import pandas as pd

data = {
    "Name": ["A", "B", "C", "D", "E", "F"],
    "Age": [18, 19, 18, 20, 19, 18],
    "Marks": [72, 85, 91, 68, 79, 95]
}

df = pd.DataFrame(data)

# 1. First 2 rows using head()
print("1. First 2 rows:\n", df.head(2))

# 2. Last 2 rows using tail()
print("\n2. Last 2 rows:\n", df.tail(2))

# 3. Number of rows and columns using shape
print("\n3. Shape:", df.shape)

# 4. Column names using columns
print("\n4. Columns:", df.columns)

# 5. Data types using dtypes
print("\n5. Data types:\n", df.dtypes)

# 6. Statistical summary using describe()
print("\n6. Statistical summary:\n", df.describe())
```
