```python
#NumPy Mathematical Functions – Find the Correct Function
"""
Consider:
import numpy as np
marks = np.array([45, 72, 89, 56, 95, 68])
Write NumPy code to display:
1. The average of the marks.
2. The position (index) of the highest mark.
3. The smallest mark.
Use appropriate NumPy functions.
"""
import numpy as np

marks = np.array([45, 72, 89, 56, 95, 68])
average = marks.sum()/len(marks)
highest_index = np.argmax(marks)
min_value= np.min(marks)
print("Average:",average)
print("Index of maximum value:",highest_index)
print("Minimum value:",min_value)
```

```python
#Comparing Array Elements
"""
Consider:
arr = np.array([10, 25, 30, 45, 50])
Write NumPy statements to display:
1. Elements greater than 30
2. Elements less than 30
3. Elements equal to 30
"""
import numpy as np


arr = np.array([10, 25, 30, 45, 50])

print("Elements greater than 30:",arr[arr>30])
print("Elements less than 30:",arr[arr<30])
print("Elements equal to 30:",arr[arr==30])
```

```python
#Processing a 2D Array Using sum()
"""
Consider:
arr = np.array([
[10, 20, 30],
[40, 50, 60],
[70, 80, 90]
])
Write NumPy code to display:
1. The sum of all elements.
2. The sum of each column using axis=0.
3. The sum of each row using axis=1.
"""
import numpy as np

arr = np.array([
[10, 20, 30],
[40, 50, 60],
[70, 80, 90]
])

print("Sum of all elements:",np.sum(arr))
print("Sum of each column (axis=0):", np.sum(arr, axis=0))
print("Sum of each row (axis=1):", np.sum(arr, axis=1))
```

```python
#Positive Index Slicing"
"""
Consider:
arr = np.array([10, 20, 30, 40, 50, 60, 70])
Write slicing statements to display:
1. Elements from index 2 to the end.
2. Elements from the beginning up to index 4
3. Elements from index 1 to index 5.
"""
import numpy as np

arr = np.array([10,20,30,40,50,60,70])
print("Index 2 to end:", arr[2:])
print("Beginning up to index 4:", arr[:4])
print("Index 1 to index 5:", arr[1:5])
```

```python
#Negative Index Slicing
"""
Consider:
arr = np.array([10, 20, 30, 40, 50, 60, 70])
Using negative indexing, write slicing statements to display:
1. The last 3 elements.
2. All elements except the last 2 elements.
3. The elements from the 5th-last position up to the 2nd-last position.
"""
import numpy as np

arr = np.array([10, 20, 30, 40, 50, 60, 70])

print("Last 3 elements:", arr[-3:])
print("All except last 2 elements:", arr[:-2])
print("5th-last to 2nd-last position:", arr[-5:-1])
```

```python
#Slicing with Step
"""
Consider:
arr = np.array([10, 20, 30, 40, 50, 60, 70, 80])
Write NumPy slicing statements to:
1. Display every second element starting from the first element.
2. Display every second element starting from the second element.
3. Reverse the complete array using slicing.
"""
import numpy as np

arr = np.array([10, 20, 30, 40, 50, 60, 70, 80])

print("Every 2nd element from 1st element:", arr[::2])
print("Every 2nd element from 2nd element:", arr[1::2])
print("Reversed array:", arr[::-1])
```

```python
#Mixed Operations
"""
Consider:
arr = np.array([15, 40, 25, 80, 35, 60, 10])
Write NumPy code to:
1. Find the maximum value.
2. Find the index of the minimum value.
3. Display only the values greater than 30.
4. Display the last three elements using negative indexing.
"""
import numpy as np

arr = np.array([15, 40, 25, 80, 35, 60, 10])

print("Maximum value:", np.max(arr))
print("Index of minimum value:", np.argmin(arr))
print("Values greater than 30:", arr[arr > 30])
print("Last three elements:", arr[-3:])
```
